### Finding All Links

```python
import requests
from bs4 import BeautifulSoup

url = "https://tinman.cs.gsu.edu/~raj/past2.html"  # put your target URL here
response = requests.get(url, verify=False)
soup = BeautifulSoup(response.text, "html.parser")

# Find all <a> tags
links = soup.find_all('a')

# Loop through and print the href attribute of each
for link in links:
    href = link.get('href')
    if href:
        print(href)
```

### Getting All Link Titles

```python
# Find all links on the page
links = soup.find_all('a')

for link in links:
    href = link.get('href')
    if href:
        # Build the full URL in case it's a relative link
        full_url = urljoin(url, href)
        print(f"Going into: {full_url}")

        # Make a request to that URL
        try:
            r = requests.get(full_url)
            if r.status_code == 200:
                # Parse the linked page
                linked_soup = BeautifulSoup(r.text, "html5lib")
                
                # Example: print the title of the linked page
                page_title = linked_soup.title.string if linked_soup.title else "No title"


                print(f"Title: {page_title}\n")
            else:
                print(f"Failed to load page, status code: {r.status_code}\n")
        except Exception as e:
            print(f"Error accessing {full_url}: {e}\n")
```

### Finding Links with Keywords


```python
# Find all links on the page
links = soup.find_all('a')

for link in links:
    href = link.get('href')
    if href:
        # Build the full URL in case it's a relative link
        full_url = urljoin(url, href)
        print(f"Going into: {full_url}")

        # Make a request to that URL
        try:
            r = requests.get(full_url)
            if r.status_code == 200:
                # Parse the linked page
                linked_soup = BeautifulSoup(r.text, "html5lib")
                
                # Example: print the title of the linked page
                page_title = linked_soup.title.string if linked_soup.title else "No title"
                print(f"Title: {page_title}\n")

                all_links = linked_soup.find_all('a', string=re.compile("Syllabus"))
                if len(all_links) == 0:
                    print('No Syllabus Found')
                else:
                    for link in all_links:
                        print(link.string, ':', link['href'])
            else:
                print(f"Failed to load page, status code: {r.status_code}\n")
        except Exception as e:
            print(f"Error accessing {full_url}: {e}\n")
        print('---------')
```

### Find Links with Variable Names

```python
# Find all links on the page
links = soup.find_all('a')

keyword = input("Enter desired link keyword:")

for link in links:
    href = link.get('href')
    if href:
        # Build the full URL in case it's a relative link
        full_url = urljoin(url, href)
        print(f"Going into: {full_url}")

        # Make a request to that URL
        try:
            r = requests.get(full_url)
            if r.status_code == 200:
                # Parse the linked page
                linked_soup = BeautifulSoup(r.text, "html5lib")
                
                # Example: print the title of the linked page
                page_title = linked_soup.title.string if linked_soup.title else "No title"
                print(f"Title: {page_title}\n")

                all_links = linked_soup.find_all('a', string=re.compile(str(keyword)))
                if len(all_links) == 0:
                    print('No Such Link Found')
                else:
                    for link in all_links:
                        print(link.string, ':', link['href'])
            else:
                print(f"Failed to load page, status code: {r.status_code}\n")
        except Exception as e:
            print(f"Error accessing {full_url}: {e}\n")
        print('---------')
```

### Print Out Content Within Links

```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import re

url = "https://tinman.cs.gsu.edu/~raj/past2.html"
keyword = input("Enter desired link keyword: ")

# Get the main page
response = requests.get(url, verify=False)
soup = BeautifulSoup(response.text, "html.parser")

# Go through each link on the main page
for link in soup.find_all("a", href=True):
    first_url = urljoin(url, link["href"])
    print(f"\nVisiting: {first_url}")
    
    try:
        # Visit the first link
        page = requests.get(first_url)
        linked_soup = BeautifulSoup(page.text, "html.parser")

        # Search for keyword-matching links
        for sublink in linked_soup.find_all("a", string=re.compile(keyword, re.IGNORECASE)):
            if not first_url.endswith('/'):
                first_url += '/'
            sub_url = urljoin(first_url, sublink["href"])
            print(f"  Found match: {sublink.string} → {sub_url}")

            sublink_href = str(sublink['href'])
            if sublink_href.endswith('.pdf'):
                print("  PDF link Found:", sublink_href)
                
            else:
                # Visit the matched link and extract head
                subpage = requests.get(sub_url)
                sub_soup = BeautifulSoup(subpage.text, "html.parser")
                print("  Title of Page: " + sub_soup.title.string)

    except Exception as e:
        print("  Error:", e)
```
